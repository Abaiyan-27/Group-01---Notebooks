# Time of Collision distribution

This notebook visualizes the class balance in the `Time of Collision` column.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Dataset.csv')
time_counts = df['Time of Collision'].fillna('Missing').value_counts()

plt.figure(figsize=(10, 5))
time_counts.sort_values().plot(kind='barh', color='#2A6F97', edgecolor='black')
plt.title('Time of Collision: Class Distribution')
plt.xlabel('Number of records')
plt.ylabel('Time of Collision')
plt.tight_layout()
plt.show()

time_counts
```

# Days Stayed on Ward

This section combines the first, second, and third ward stay columns into one view so you can compare ward stay duration across all ward types.

```python
import numpy as np

ward_columns = {
    'First Ward': 'Number of days in first ward',
    'Second Ward': 'Number of days in Second ward',
    'Third Ward': 'Number of days in Third ward',
}

ward_days = (
    df[list(ward_columns.values())]
    .rename(columns={v: k for k, v in ward_columns.items()})
    .melt(var_name='Ward Type', value_name='Days Stayed')
)

ward_days['Days Stayed'] = pd.to_numeric(ward_days['Days Stayed'], errors='coerce')
ward_days = ward_days.dropna(subset=['Days Stayed'])

ward_summary = ward_days.groupby('Ward Type')['Days Stayed'].agg(['count', 'mean', 'median', 'sum']).sort_values('mean', ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(ward_summary.index, ward_summary['mean'], color=['#2A6F97', '#C75146', '#6A994E'], edgecolor='black')
plt.title('Average Number of Days Stayed by Ward Type')
plt.xlabel('Ward Type')
plt.ylabel('Average days stayed')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()

ward_summary
```
